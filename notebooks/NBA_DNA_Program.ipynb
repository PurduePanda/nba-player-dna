```python
# Starting with Luka and SGA

scaled_player_features[
    scaled_player_features["Player"].isin(
        ["Luka Dončić", "Shai Gilgeous-Alexander"]
    )
]
```

```python
luka_vector = scaled_player_features.loc[scaled_player_features["Player"] == "Luka Dončić", candidate_features].iloc[0]
shai_vector = scaled_player_features.loc[scaled_player_features["Player"] == "Shai Gilgeous-Alexander", candidate_features].iloc[0]

luka_vector.shape
shai_vector.shape
```

```python
dot_product = np.dot(luka_vector, shai_vector)
luka_norm = np.linalg.norm(luka_vector)
shai_norm = np.linalg.norm(shai_vector)

cosine_similarity = dot_product / (luka_norm * shai_norm)
print(cosine_similarity)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

luka_2d = luka_vector.to_numpy().reshape(1, -1)
shai_2d = shai_vector.to_numpy().reshape(1, -1)

sklearn_similarity = cosine_similarity(luka_2d, shai_2d)
print(sklearn_similarity)
score = sklearn_similarity[0][0]
print(score)
```

## Creating the compare_players() Function to Compare Any Two Players Given Input

```python
def compare_players(player_1, player_2):
  player1_vector = scaled_player_features.loc[scaled_player_features["Player"] == player_1, candidate_features].iloc[0]
  player2_vector = scaled_player_features.loc[scaled_player_features["Player"] == player_2, candidate_features].iloc[0]

  player1_2d = player1_vector.to_numpy().reshape(1, -1)
  player2_2d = player2_vector.to_numpy().reshape(1, -1)

  sklearn_similarity = cosine_similarity(player1_2d, player2_2d)
  score = sklearn_similarity[0][0]

  return score
```

## Making Reusable Function find_similar_players()

```python
def find_similar_players(player, n):

  if player not in scaled_player_features["Player"].values:
    raise ValueError(f"Player {player} was not found in the dataset.")

  player_vector = scaled_player_features.loc[
    scaled_player_features["Player"] == player,
    candidate_features
  ].to_numpy()

  all_similarities = cosine_similarity(
    player_vector,
    feature_matrix
  )

  similarity_results = pd.DataFrame({
    "Player": scaled_player_features["Player"],
    "Similarity": all_similarities[0]
  })
  similarity_results = similarity_results.sort_values(
    by="Similarity",
    ascending=False
  )
  similarity_results = similarity_results[
    similarity_results["Player"] != player
  ]
  similarity_results = similarity_results.reset_index(drop=True)
  similarity_results.index = similarity_results.index + 1

  return similarity_results.head(n)
```

```python
# removing redundant features, creating features version 2

v2_features = [
    "FGA_36",
    "3PA_RATE",
    "FTA_RATE",
    "TS%",
    "AST_36",
    "TOV_36",
    "REB_36",
    "STL_36",
    "BLK_36"
]
```

```python
v2_scaled = pd.DataFrame(
    v2_scaled_array,
    columns=v2_features,
    index=player_features.index
)

v2_scaled.insert(
    0,
    "Player",
    player_features["Player"]
)
```

```python
v2_feature_matrix = v2_scaled[v2_features].to_numpy()

print(v2_feature_matrix.shape)
```

```python
# importing both distance metrics we used before

from sklearn.metrics.pairwise import cosine_similarity
from sklearn.metrics.pairwise import euclidean_distances
```

```python
def find_neighbors(player, metric="cosine", n=10):
  if player not in v2_scaled["Player"].values:
    raise ValueError(
        f"Player {player} was not found in the dataset."
    )

  player_vector = v2_scaled.loc[
      v2_scaled["Player"] == player,
      v2_features
  ].to_numpy()

  if metric == "cosine":
    all_distances = cosine_similarity(
        player_vector,
        v2_feature_matrix
    ).flatten() # flattened to 1D to prevent DataFrame construction errors
    results = pd.DataFrame({
        "Player": v2_scaled["Player"],
        "Distance": all_distances
    })
    results = results[
        results["Player"] != player
    ]
    results = results.sort_values(
        by="Distance",
        ascending=False
    )
  elif metric == "euclidean":
    all_distances = euclidean_distances(
        player_vector,
        v2_feature_matrix
    ).flatten() # flattened to 1D to prevent DataFrame construction errors
    results = pd.DataFrame({
        "Player": v2_scaled["Player"],
        "Distance": all_distances
    })
    results = results[
        results["Player"] != player
    ]
    results = results.sort_values(
        by="Distance",
        ascending=True
    )
  else:
    raise ValueError(
        f"Metric {metric} is not supported."
    )

  results = results.head(n).reset_index(drop=True)

  results.index += 1

  return results
```

# **Week 4: Freeze the V2 Euclidean Model and Produce One Clean Application-Ready JSON Dataset**

```python
# compute the entire distance matrix

from sklearn.metrics.pairwise import euclidean_distances

feature_matrix = v2_scaled[v2_features].to_numpy()

distance_matrix = euclidean_distances(
    feature_matrix,
    feature_matrix
)

print(distance_matrix.shape)
```

```python
distance_matrix[:5, :5]
```

```python
# precompute the top 10 neighbors

import numpy as np

top_neighbors = {}

for i, player in enumerate(v2_scaled["Player"]):
  sorted_indices = np.argsort(distance_matrix[i])

  # index 0 will be the player themselves
  neighbor_indices = sorted_indices[1:11]

  neighbors = []

  for j in neighbor_indices:
    neighbors.append({
        "player": v2_scaled.iloc[j]["Player"],
        "distance": float(distance_matrix[i][j])
    })

  top_neighbors[player] = neighbors
```

```python
top_neighbors["Stephen Curry"]
```

```python
# adding basic player metadata

player_metadata = eligible_players[
    ["Player", "Team", "Age", "Min"]
].copy()
```

```python
player_metadata.head()
```

```python
# building one app-ready player record

player_name = "Stephen Curry"

metadata = player_metadata[
    player_metadata["Player"] == player_name
].iloc[0]

raw_features = v2_unscaled[
    v2_unscaled["Player"] == player_name
].iloc[0]

scaled_features = v2_scaled[
    v2_scaled["Player"] == player_name
].iloc[0]
```

```python
# generate every player's record

players_output = []

for player_name in v2_scaled["Player"]:
  metadata = player_metadata[
      player_metadata["Player"] == player_name
  ].iloc[0]

  raw_row = v2_unscaled[
      v2_unscaled["Player"] == player_name
  ].iloc[0]

  scaled_row = v2_scaled[
      v2_scaled["Player"] == player_name
  ].iloc[0]

  raw_features = {
      feature: float(raw_row[feature])
      for feature in v2_features
  }

  scaled_features = {
      feature: float(scaled_row[feature])
      for feature in v2_features
  }

  player_record = {
      "name": player_name,
      "team": str(metadata["Team"]),
      "age": float(metadata["Age"]),
      "minutes": float(metadata["Min"]),
      "features": raw_features,
      "scaled_features": scaled_features,
      "neighbors": top_neighbors[player_name]
  }

  players_output.append(player_record)
```

```python
players_output[0]
```

```python
# add model metadata

model_metadata = {
    "season": "2025-26",
    "minimum_minutes": 500,
    "metric": "euclidean distance",
    "features": v2_features,
    "number_of_players": len(v2_scaled),
    "scaler_mean": {
        feature: float(mean)
        for feature, mean
        in zip(v2_features, scaler.mean_)
    },
    "scaler_scale": {
        feature: float(scale)
        for feature, scale
        in zip(v2_features, scaler.scale_)
    }
}
```

```python
# construct the complete application dataset

app_data = {
    "metadata": model_metadata,
    "players": players_output
}
```
